```python
def sense(DNA):
    RNA=''
    for nukl in DNA.upper()[:21]:
        if nukl=='T':
            RNA+='U'
        else:
            RNA+=nukl
    return RNA

def guide_qq(RNA):
    guide_siRNA=''
    for nukle in RNA:
        if nukle=='C':
            guide_siRNA+='G'
        elif nukle=='U':
            guide_siRNA+='A'
        elif nukle=='A':
            guide_siRNA+='U'
        elif nukle=='G':
            guide_siRNA+='C'
        else:
            print('wrong sequence')
    return guide_siRNA

def passenger(DNA):
    passenger=''
    for nucl in DNA.upper()[2:]:
        if nucl=="T":
            passenger+="U"
        else:
            passenger+=nucl
    return passenger

def yi_tel(guide_siRNA,passenger):
    eff=0
    for nukl in guide_siRNA[0]:
        if nukl=='A' or nukl=='U':
            eff+=25
    AU_counter=0
    for nukl in guide_siRNA[:7]:
        if nukl=='A' or nukl=='U':
            AU_counter+=1
    if (AU_counter/7)*100>50:
        eff+=25
    gc_counter=0
    for nukl in guide_siRNA[0]:
        if nukl=='G' or nukl=='C':
            gc_counter+=1
        else:
            gc_counter=0
        if gc_counter>10:
            break
        if gc_counter<=10:
            eff+=25
    return eff

def rev_complement(RNA):
    rev_compl = ''
    for nukl in RNA:
        if nukl=='A':
            nukl+='U'
        if nukl=='G':
            nukl+='C'
        if nukl=='C':
            nukl+='G'
        if nukl=='U':
            nukl+='A'
    return rev_compl[::-1]

def Rainolds(guide_siRNA,passenger):
    efff=0
    c_1R=0
    for nukl in guide_siRNA:
        if nukl=='G' or nukl=='C':
            c_1R+=1
    if (c_1R/21)*100>30 and (c_1R/21)*100<52:
        efff+=100/7
    c_nefor=0
    for nukl in passenger[:5]:
        if nukl=='A' or nukl=='U':
            c_nefor+=1
    if c_nefor>=0.6:
        efff+=100/7
    if passenger[::-1][2]=='A':
        efff+=100/7
    if passenger[::-1][9]=='U':
        efff+=100/7
    if passenger[::-1][18]=='A' or passenger[18]=='U':
        efff+=100/7
    if passenger[::-1][12]!='G':
        efff+=100/7
    ccounter=0
    for i in range(7):
        for k in range(6,10):
            for j in range(2,10):
                if i+k+j+k-1<len(guide_siRNA):
                    if guide_siRNA[j:i+k] == rev_complement(guide_siRNA[j+k+j:j+k+j+k]):
                        ccounter+=1
    if ccounter==0:
        efff+=100/7
    return efff

def Amartzguiui(guide_siRNA,passenger):
    e_ff=0
    if guide_siRNA[18]=='C' or guide_siRNA[18]=='G':
        e_ff+=100/6
    if guide_siRNA[0]=='A' or guide_siRNA[0]=='U':
        e_ff+=100/6
    if guide_siRNA[13]=='A' or guide_siRNA[13]=='U':
        e_ff+=100/6
    if guide_siRNA[18]!='U':
        e_ff+=100/6
    if guide_siRNA[0]!='G':
        e_ff+=100/6
    au_counter=0
    counter = 0
    for nukl in guide_siRNA[:3]:
        if nukl=='U' or nukl=='A':
            au_counter+=1
    if au_counter>=2:
        counter += 1
    gc_counter=0
    for nukl in guide_siRNA[18:]:
        if nukl=='G' or nukl=='C':
            gc_counter+=1
    if gc_counter>=2:
        counter+=1
    if counter==2:
        e_ff+=100/6
    return e_ff

def sred(eff,efff,e_ff):
    Eff_av=0
    if (eff+efff+e_ff)/3>=0:
        Eff_av+=(eff+efff+e_ff)/3
    return Eff_av

DNA_list=[]
eff_list=[]

DNA_seq=input()
for i in range(len(DNA_seq)-23):
    DNA=DNA_seq[i:i+23]
    RNA=sense(DNA)
    guide_siRNA=guide_qq(RNA)
    passenger_siRNA=passenger(DNA)
    eff=yi_tel(guide_siRNA, passenger_siRNA)
    efff=Rainolds(guide_siRNA, passenger_siRNA)
    e_ff=Amartzguiui(guide_siRNA,passenger_siRNA)
    Eff_av=sred(eff,efff,e_ff)
    DNA_list.append(DNA_seq)
    eff_list.append(Eff_av)
    Eff_max=max(eff_list)
for i in range (len(eff_list)):
    if eff_list[i]==Eff_max:
        DNA_seq=DNA_list[i]
        RNA=sense(DNA)
        guide_siRNA=guide_qq(RNA)
        passenger_siRNA=passenger(DNA)
        eff=yi_tel(guide_siRNA, passenger_siRNA)
        efff=Rainolds(guide_siRNA, passenger_siRNA)
        e_ff=Amartzguiui(guide_siRNA,passenger_siRNA)
        Eff_av=sred(eff,efff,e_ff)
        print("DNA, 5'-3':",DNA.upper())
        print("mRNA, 5'-3':",RNA)
        print("guide_siRNA, 5'-3':",guide_siRNA)
        print("passenger, 3'-5':",passenger_siRNA)
        print('efficiency_1:',eff,'%')
        print('efficiency_2:',efff,'%')
        print('efficiency_3:',e_ff,'%')
        print('Averfge efficiency:',Eff_av,'%')
        print('Maximum efficiency:',Eff_max,'%')
```
